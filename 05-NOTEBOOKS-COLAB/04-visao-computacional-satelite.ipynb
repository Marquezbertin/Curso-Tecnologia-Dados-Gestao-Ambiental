## Opção B — MobileNetV2 pré-treinado

```python
try:
    import tensorflow as tf
    print("TensorFlow", tf.__version__)
except Exception as e:
    print("TensorFlow nao disponivel:", e)
```

```python
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input, decode_predictions
from tensorflow.keras.preprocessing import image
import numpy as np

modelo = MobileNetV2(weights="imagenet")
print("modelo carregado")
```

## Passo 1 — Enviar uma imagem

No Colab, rode a célula e escolha uma foto do pátio, de uma árvore, de um rio ou de uma área
queimada. **Não envie foto com rosto de pessoa.**

```python
from google.colab import files
enviados = files.upload()
```

## Passo 2 — Classificar

```python
nome_arquivo = list(enviados.keys())[0]
img = image.load_img(nome_arquivo, target_size=(224, 224))
x = image.img_to_array(img)
x = np.expand_dims(x, axis=0)
x = preprocess_input(x)

predicoes = modelo.predict(x)
for classe, descricao, prob in decode_predictions(predicoes, top=5)[0]:
    print(f"{descricao:30s} {prob:.3f}")
```

## O que observar

- O modelo foi treinado em **imagens comuns** (ImageNet), não em imagens de satélite.
- Ele pode confundir fumaça com nuvem, pasto com floresta, telhado com água.
- A probabilidade alta não significa certeza: significa confiança do modelo nos dados dele.
- Para uso ambiental sério, seria necessário treinar com imagens da região e validar com
  especialistas.
