## Passo 1 — Importar as bibliotecas

```python
import pandas as pd
import matplotlib.pyplot as plt

print("pandas", pd.__version__)
print("matplotlib", plt.matplotlib.__version__)
```

## Passo 2 — Ler o arquivo

No Colab: clique no ícone de pasta (à esquerda), faça upload de
`06-DADOS/exemplos/exemplo-inmet-sintetico.csv` e depois rode a célula.

```python
caminho = "exemplo-inmet-sintetico.csv"
df = pd.read_csv(caminho)
df.head()
```

## Passo 3 — Conhecer a tabela

```python
df.info()
```

```python
df.describe()
```

## Passo 4 — Limpar

Vamos converter a data, remover vazios e ordenar. Em dados reais do INMET, o código `-9999`
costuma indicar falha de medição.

```python
df["data"] = pd.to_datetime(df["data"])
df = df.dropna()
df = df[df["temp_max_c"] > -100]
df = df.sort_values("data").reset_index(drop=True)
df.head()
```

## Passo 5 — Primeiro gráfico

```python
plt.figure(figsize=(10, 4))
plt.plot(df["data"], df["temp_max_c"], label="Temperatura máxima")
plt.plot(df["data"], df["temp_min_c"], label="Temperatura mínima")
plt.title("Temperatura diária — dados sintéticos didáticos")
plt.xlabel("Data")
plt.ylabel("Temperatura (°C)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 3))
plt.bar(df["data"], df["chuva_mm"], color="steelblue")
plt.title("Chuva diária — dados sintéticos didáticos")
plt.xlabel("Data")
plt.ylabel("Chuva (mm)")
plt.grid(axis="y")
plt.tight_layout()
plt.show()
```
