## Passo 1 — Instalar e importar

```python
!pip -q install folium
import folium

print("folium", folium.__version__)
```

## Passo 2 — Centro do mapa

Substitua `lat` e `lon` pelas coordenadas do seu município (clique com o botão direito no Google
Maps e copie os números). O exemplo abaixo usa uma coordenada aproximada do interior de São Paulo.

```python
lat, lon = -22.55, -47.90
mapa = folium.Map(location=[lat, lon], zoom_start=12, tiles="OpenStreetMap")
mapa
```

## Passo 3 — Adicionar pontos

Troque os nomes e as coordenadas pelos pontos do seu projeto: escola, nascente, ponto de coleta,
área de descarte.

```python
pontos = [
    {"nome": "Escola", "lat": -22.550, "lon": -47.900, "tipo": "coleta"},
    {"nome": "Ponto de coleta 1", "lat": -22.555, "lon": -47.905, "tipo": "coleta"},
    {"nome": "Area verde", "lat": -22.545, "lon": -47.895, "tipo": "verde"},
]

for p in pontos:
    folium.Marker(
        [p["lat"], p["lon"]],
        popup=f"{p['nome']} ({p['tipo']})",
        tooltip=p["nome"],
    ).add_to(mapa)

mapa
```

## Passo 4 — Camada de círculos

Círculos ajudam a mostrar intensidade ou área de influência. O raio está em metros.

```python
for p in pontos:
    folium.Circle(
        [p["lat"], p["lon"]],
        radius=300,
        color="green" if p["tipo"] == "verde" else "blue",
        fill=True,
        fill_opacity=0.2,
        popup=p["nome"],
    ).add_to(mapa)

mapa
```

## Passo 5 — Controle de camadas e exportação

```python
folium.LayerControl().add_to(mapa)
mapa.save("mapa-do-municipio.html")
mapa
```

## Passo 6 — Legenda e interpretação
